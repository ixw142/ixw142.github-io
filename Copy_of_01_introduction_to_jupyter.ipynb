## Exploring Math with Sliders

Let's use sliders to explore addition interactively. Move the sliders and watch the result change!

```python
from ipywidgets import IntSlider, interact


def add_numbers(a=5, b=3):
    """Add two numbers and display the result"""
    result = a + b
    print(f"{a} + {b} = {result}")
    print("=" * 30)
    print("💡 Try different values to see how addition works!")


interact(
    add_numbers,
    a=IntSlider(min=0, max=20, step=1, value=5, description="First number:"),
    b=IntSlider(min=0, max=20, step=1, value=3, description="Second number:"),
);
```

## Exploring Different Operations

Python can do many math operations. Let's explore them interactively!

```python
from ipywidgets import Dropdown, IntSlider, interact


def calculate(a=10, b=5, operation="+"):
    """Perform different math operations"""
    if operation == "+":
        result = a + b
        op_name = "Addition"
    elif operation == "-":
        result = a - b
        op_name = "Subtraction"
    elif operation == "*":
        result = a * b
        op_name = "Multiplication"
    elif operation == "/":
        result = a / b if b != 0 else "Cannot divide by zero!"
        op_name = "Division"

    print(f"Operation: {op_name}")
    print(f"{a} {operation} {b} = {result}")
    print("=" * 40)


interact(
    calculate,
    a=IntSlider(min=0, max=50, step=1, value=10, description="First:"),
    b=IntSlider(min=0, max=50, step=1, value=5, description="Second:"),
    operation=Dropdown(
        options=["+", "-", "*", "/"], value="+", description="Operation:"
    ),
);
```

## Visualizing Data with Plots

Python is excellent for creating visualizations. Let's create an interactive plot that changes based on your input!

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import Dropdown, FloatSlider, interact


def plot_function(function="Linear", slope=1.0, intercept=0.0):
    """Plot different mathematical functions"""
    x = np.linspace(-10, 10, 100)

    if function == "Linear":
        y = slope * x + intercept
        title = f"y = {slope:.1f}x + {intercept:.1f}"
    elif function == "Quadratic":
        y = slope * x**2 + intercept
        title = f"y = {slope:.1f}x² + {intercept:.1f}"
    elif function == "Cubic":
        y = slope * x**3 + intercept
        title = f"y = {slope:.1f}x³ + {intercept:.1f}"

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, linewidth=2, color="blue")
    plt.grid(True, alpha=0.3)
    plt.axhline(y=0, color="black", linewidth=0.5)
    plt.axvline(x=0, color="black", linewidth=0.5)
    plt.title(title, fontsize=14, fontweight="bold")
    plt.xlabel("x", fontsize=12)
    plt.ylabel("y", fontsize=12)
    plt.ylim(-50, 50)
    plt.show()


interact(
    plot_function,
    function=Dropdown(
        options=["Linear", "Quadratic", "Cubic"],
        value="Linear",
        description="Function:",
    ),
    slope=FloatSlider(min=-5, max=5, step=0.1, value=1.0, description="Slope:"),
    intercept=FloatSlider(
        min=-10, max=10, step=0.5, value=0.0, description="Intercept:"
    ),
);
```

## Interactive Color Mixer

Let's explore how colors work by mixing Red, Green, and Blue (RGB) values!

```python
import matplotlib.patches as patches
from ipywidgets import IntSlider, interact


def mix_colors(red=255, green=128, blue=0):
    """Mix RGB colors and display the result"""
    # Normalize values to 0-1 range for matplotlib
    r_norm = red / 255
    g_norm = green / 255
    b_norm = blue / 255

    # Create a figure to show the color
    fig, ax = plt.subplots(figsize=(8, 4))

    # Create a colored rectangle
    rect = patches.Rectangle((0, 0), 1, 1, facecolor=(r_norm, g_norm, b_norm))
    ax.add_patch(rect)

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")

    # Display RGB values
    color_code = f"RGB({red}, {green}, {blue})"
    hex_code = f"#{red:02x}{green:02x}{blue:02x}".upper()

    # Choose text color based on brightness
    brightness = (red + green + blue) / 3
    text_color = "white" if brightness < 128 else "black"

    ax.text(
        0.5,
        0.6,
        color_code,
        ha="center",
        va="center",
        fontsize=16,
        fontweight="bold",
        color=text_color,
    )
    ax.text(0.5, 0.4, hex_code, ha="center", va="center", fontsize=14, color=text_color)

    plt.title("Your Mixed Color", fontsize=14, pad=10)
    plt.tight_layout()
    plt.show()

    print(f"🎨 RGB: ({red}, {green}, {blue})")
    print(f"🔖 Hex: {hex_code}")


interact(
    mix_colors,
    red=IntSlider(min=0, max=255, step=1, value=255, description="Red:"),
    green=IntSlider(min=0, max=255, step=1, value=128, description="Green:"),
    blue=IntSlider(min=0, max=255, step=1, value=0, description="Blue:"),
);
```

## Text Input Widget

Let's create a fun text-based interaction. Type your message and click the button!

```python
from ipywidgets import Output, Text

# Create widgets
text_input = Text(
    value="",
    placeholder="Type something here...",
    description="Message:",
    disabled=False,
)

button = Button(
    description="Transform!",
    button_style="success",
    tooltip="Click to transform your text",
)

output = Output()


def on_button_click(b):
    """Transform the input text in various ways"""
    with output:
        output.clear_output()
        message = text_input.value

        if message:
            print("Original:", message)
            print("Uppercase:", message.upper())
            print("Lowercase:", message.lower())
            print("Reversed:", message[::-1])
            print("Length:", len(message), "characters")
            print("Word count:", len(message.split()), "words")
        else:
            print("Please enter some text first!")


button.on_click(on_button_click)

# Display widgets
display(VBox([text_input, button, output]))
```

## Interactive Pattern Generator

Let's create some ASCII art patterns! Adjust the size and character to see different patterns.

```python
from ipywidgets import Dropdown, IntSlider, interact


def create_pattern(size=5, pattern_type="Square", character="*"):
    """Generate different ASCII patterns"""
    print(f"Pattern Type: {pattern_type}")
    print("=" * 40)

    if pattern_type == "Square":
        for i in range(size):
            print(character * size)

    elif pattern_type == "Triangle":
        for i in range(1, size + 1):
            print(character * i)

    elif pattern_type == "Pyramid":
        for i in range(1, size + 1):
            spaces = " " * (size - i)
            stars = character * (2 * i - 1)
            print(spaces + stars)

    elif pattern_type == "Diamond":
        # Top half
        for i in range(1, size + 1):
            spaces = " " * (size - i)
            stars = character * (2 * i - 1)
            print(spaces + stars)
        # Bottom half
        for i in range(size - 1, 0, -1):
            spaces = " " * (size - i)
            stars = character * (2 * i - 1)
            print(spaces + stars)

    elif pattern_type == "Hollow Square":
        for i in range(size):
            if i == 0 or i == size - 1:
                print(character * size)
            else:
                print(character + " " * (size - 2) + character)


interact(
    create_pattern,
    size=IntSlider(min=3, max=15, step=1, value=5, description="Size:"),
    pattern_type=Dropdown(
        options=["Square", "Triangle", "Pyramid", "Diamond", "Hollow Square"],
        value="Square",
        description="Pattern:",
    ),
    character=Dropdown(
        options=["*", "#", "@", "+", "◆", "●", "■"], value="*", description="Character:"
    ),
);
```

## Temperature Converter

Let's build a practical tool - a temperature converter that converts between Celsius and Fahrenheit!

```python
from ipywidgets import Dropdown, FloatSlider, interact


def convert_temperature(value=0.0, convert_from="Celsius"):
    """Convert temperature between Celsius and Fahrenheit"""

    if convert_from == "Celsius":
        fahrenheit = (value * 9 / 5) + 32
        print(f"🌡️  {value:.1f}°C = {fahrenheit:.1f}°F")

        # Add some context
        if value <= 0:
            print("❄️  Freezing or below!")
        elif value < 15:
            print("🧊  Cold")
        elif value < 25:
            print("😊  Comfortable")
        elif value < 35:
            print("☀️  Warm")
        else:
            print("🔥  Hot!")

    else:  # Fahrenheit to Celsius
        celsius = (value - 32) * 5 / 9
        print(f"🌡️  {value:.1f}°F = {celsius:.1f}°C")

        # Add some context
        if celsius <= 0:
            print("❄️  Freezing or below!")
        elif celsius < 15:
            print("🧊  Cold")
        elif celsius < 25:
            print("😊  Comfortable")
        elif celsius < 35:
            print("☀️  Warm")
        else:
            print("🔥  Hot!")

    print("\n💡 Fun Facts:")
    print("   • Water freezes at 0°C (32°F)")
    print("   • Water boils at 100°C (212°F)")
    print("   • Normal body temperature is 37°C (98.6°F)")


interact(
    convert_temperature,
    value=FloatSlider(
        min=-50, max=150, step=0.5, value=0.0, description="Temperature:"
    ),
    convert_from=Dropdown(
        options=["Celsius", "Fahrenheit"], value="Celsius", description="Convert from:"
    ),
);
```

## Jupyter Keyboard Shortcuts

Here are some helpful keyboard shortcuts to make you more efficient:

### Command Mode (press Esc to enter)
- **A**: Insert cell above
- **B**: Insert cell below
- **DD**: Delete cell (press D twice)
- **M**: Change cell to Markdown
- **Y**: Change cell to Code
- **Z**: Undo cell deletion

### Edit Mode (press Enter to enter)
- **Shift + Enter**: Run cell and move to next
- **Ctrl + Enter**: Run cell and stay in place
- **Alt + Enter**: Run cell and insert new cell below
- **Tab**: Code completion/indent
- **Shift + Tab**: Show function help

Try these out! Press **Esc** to enter Command Mode, then press **B** to create a new cell below this one.

## Tips for Success with Jupyter

1. **Run cells in order**: Jupyter notebooks work best when you run cells from top to bottom
2. **Save often**: Click the save button or press Ctrl+S (Cmd+S on Mac)
3. **Restart if confused**: If things aren't working, try "Kernel > Restart & Run All" from the menu
4. **Experiment freely**: You can't break anything! Try changing code and see what happens
5. **Read error messages**: They often tell you exactly what's wrong
6. **Use comments**: Add comments to explain your code (lines starting with #)
7. **Keep it organized**: Use Markdown cells to add explanations and structure
